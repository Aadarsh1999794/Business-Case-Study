```python
#Let us write a function to do the exact same thing for whichever director we want.
```

```python
def solve(director_name, start_year, end_year):
  director_id = directors[ directors['director_name'] == director_name].reset_index(drop = True)["id"][0]

  director_movies = movies[(movies["director_id"] == director_id)]

  df = director_movies[ (director_movies["year"] >= start_year) & (director_movies["year"] <= end_year) ]

  return df
```

```python
solve("James Cameron", 2005, 2010)
```
