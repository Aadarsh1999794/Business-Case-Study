#### Understand the relationship

```python
df['Contract'].describe()
```

###### We want to answer: Which contract type has the highest percentage of customers leaving?

```python
pd.crosstab(df['Contract'],df['Churn'])
```

```python
churn_rate = pd.crosstab(df["Contract"],df["Churn"],normalize="index") * 100
```

```python
churn_rate
```

```python
churn_rate["Yes"].plot(kind="bar",figsize=(5, 5))
plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

###### The contract type with the highest churn rate is Month-to-month.
###### Its churn rate is approximately 42%.
###### This suggests that customers with Month-to-month contracts may be more likely to leave, so the company should investigate ways to improve retention for this group.

#### Business question: Are newer customers more likely to leave than customers who have stayed for several years?

```python
bins = [0, 12, 24, 48, 60, np.inf]
labels = ["0-12 months", "13-24 months","25-48 months", "49-60 months", "60+ months"]
```

```python
df["TenureGroup"] = pd.cut(df["tenure"],bins=bins,labels=labels,include_lowest=True)
```

```python
tenure_churn = pd.crosstab(df["TenureGroup"],df["Churn"],normalize="index") * 100
```

```python
tenure_churn
```

```python
sns.barplot(data=df,x="TenureGroup",y=df["Churn"].eq("Yes").astype(int),errorbar=None)
plt.title("Customer Churn by Tenure")
plt.xlabel("Customer Tenure")
plt.ylabel("Churn Rate")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

##### Does monthly billing relate to churn?

```python
df.groupby("Churn")["MonthlyCharges"].mean()
```

```python
sns.boxplot(data=df,x="Churn",y="MonthlyCharges")
plt.title("Monthly Charges vs Customer Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.show()
```

#### Which payment methods are associated with higher churn?

```python
payment_churn = pd.crosstab(df["PaymentMethod"],df["Churn"],normalize="index") * 100
```

```python
payment_churn
```

```python
payment_churn["Yes"].sort_values().plot(kind="barh",figsize=(9, 5))
plt.title("Customer Churn Rate by Payment Method")
plt.xlabel("Churn Rate (%)")
plt.ylabel("Payment Method")
plt.tight_layout()
plt.show()
```

#### Which internet service customers churn the most?

```python
internet_churn = pd.crosstab(df["InternetService"],df["Churn"],normalize="index") * 100
```

```python
internet_churn
```

```python
internet_churn["Yes"].sort_values().plot(kind="bar",figsize=(8, 5))
plt.title("Customer Churn by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Final Recommendations: Customer Churn Analysis

### 1. Improve Early Customer Retention
Customers with 0–12 months of tenure have the highest churn rate at **47.44%**. The company should improve customer onboarding, provide proactive support during the first few months, and monitor customer satisfaction to identify potential churn risks early.

### 2. Investigate Electronic-Check Customers
Customers using electronic checks have a churn rate of **45.29%**, compared with 15.24% among automatic credit-card users. The company should investigate billing experiences, payment preferences, contract types, and customer profiles. It could also test whether making automatic payment options easier to use improves retention.

### 3. Investigate Fiber-Optic Customer Churn
Fiber-optic customers have a churn rate of **41.89%**, compared with 18.96% for DSL customers. The company should investigate service reliability, pricing, installation experiences, and technical support to identify potential areas for improvement.

### 4. Review Pricing and Customer Value
Customers who churned paid an average of **74.44 per month**, compared with 61.27 among customers who stayed. The company should examine whether higher charges correspond to lower perceived value, unsuitable service bundles, or other customer concerns. Targeted plan reviews may help improve retention.

### 5. Monitor Churn Performance
The overall churn rate is **26.54%**. The company should monitor churn regularly and segment results by tenure, contract type, payment method, and internet service. This will help identify emerging patterns and evaluate whether retention initiatives are effective.

### 6. Validate Recommendations Before Implementation
The findings show associations rather than proven causes of churn. Before implementing large-scale changes, the company should compare similar customer groups, investigate additional variables, and test retention strategies using measurable outcomes.

## Conclusion

The analysis suggests prioritizing early-tenure customers, electronic-check users, and fiber-optic customers for further investigation. By combining targeted customer support, billing improvements, service-quality reviews, and regular performance monitoring, the company can develop evidence-based strategies to reduce churn and improve customer retention.
